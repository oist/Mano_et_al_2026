### Plot drift trace

```python
rigid_shifts = np.load(os.path.join(rootdir, "run1/mcorr_rigid", "shifts_rig.npy"))
x_shifts = np.load(os.path.join(rootdir, "run1/mcorr_nonrigid", "y_shifts_els.npy"))
y_shifts = np.load(os.path.join(rootdir, "run1/mcorr_nonrigid", "x_shifts_els.npy"))
```

```python
fig = plt.figure(figsize=(4,1.8), dpi=300)
gs = fig.add_gridspec(2, 2, width_ratios=[1, 0.2])

# 1 pixel = 0.52 um
xs = (rigid_shifts[30:-30, 0] + x_shifts.T).T * 0.52
ys = (rigid_shifts[30:-30, 1] + y_shifts.T).T * 0.52
xs = xs[::5]
ys = ys[::5]
t = 1/15.4 * 5 * np.arange(xs.shape[0])

for i in range(2):
    if i == 0:
        s = xs
    else:
        s = ys
    ax = fig.add_subplot(gs[i, 0])
    for jj in range(x_shifts.shape[1]):
        ax.plot(t, s[:, jj], color="gray", alpha=0.25, lw=0.5)
    ax.plot(t, s.mean(axis=1), color="k", lw=0.5)
    if i == 0:
        ax.set_ylabel('X shifts ($\mathrm{\mu}$m)')
        # ax.set_xticks([])
    else:
        ax.set_ylabel('Y shifts ($\mathrm{\mu}$m)')
        ax.set_xlabel("Time (s)")
    ax.set_xlim(t[0], t[-1])
    ax.set_ylim([-24,24])
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

for i in range(2):
    if i == 0:
        s = xs
    else:
        s = ys
    ax = fig.add_subplot(gs[i, 1])
    h, bin_edges = np.histogram(s.ravel(), bins=32, density=True)
    bins = (bin_edges[:-1] + bin_edges[1:])/2
    ax.barh(bins, h, height=bin_edges[:-1]-bin_edges[1:], color="gray")
    ax.set_ylim([-24,24])
    # ax.set_yticks([])
    ax.set_xlim(0, 0.15)
    if i == 0:
        pass
    else:
        ax.set_xlabel("Counts")
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig(os.path.join(tmpdir, "drift_trace.pdf"), bbox_inches="tight")
plt.show()

print(f"X shift: Mean = {xs.ravel().mean()}, STD = {xs.ravel().std()}")
print(f"Y shift: Mean = {ys.ravel().mean()}, STD = {ys.ravel().std()}")
```
